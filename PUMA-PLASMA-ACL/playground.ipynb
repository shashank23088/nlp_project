## PREPROCESSING

```python
import json
```

```python
with open('./data/train.json', 'r') as file:
    train_json = json.load(file)

with open('./data/valid.json', 'r') as file:
    valid_json = json.load(file)

with open('./data/test.json', 'r') as file:
    test_json = json.load(file)
```

```python
def preprocess(json_file, split):

    preprocessed_json = []
    # perspectives = ['INFORMATION', 'QUESTION', 'EXPERIENCE', 'SUGGESTION', 'CAUSE']

    for entry in json_file:
        question = entry['question']
        answers = entry['answers']

        perspective_summaries = entry['labelled_summaries']
        for perspective_summary, summary in perspective_summaries.items():
            perspective = perspective_summary.split('_')[0]
            json_entry = {
                'question': question,
                'answers': answers, 
                'Perspective': perspective,
                'Summary': summary
            }

            preprocessed_json.append(json_entry)

    save_file = f'./data/{split}_preprocessed.json'

    with open(save_file, 'w') as file:
        json.dump(preprocessed_json, file, indent=4)

    print(f'preprocessed file saved at: {save_file}!')
```

```python
preprocess(train_json, 'train')
```

```python
preprocess(valid_json, 'valid')
```

```python
preprocess(test_json, 'test')
```
